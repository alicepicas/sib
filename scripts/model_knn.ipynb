```python
# imports
from si.io.csv_file import read_csv
from si.model_selection.split import train_test_split
from si.models.knn_classifier import KNNClassifier
from si.model_selection.split import stratified_train_test_split
from si.models.knn_regressor import KNNRegressor
```

```python
# datasets
iris_dataset = read_csv('../datasets/iris/iris.csv', features=True, label=True)
```

```python
# split dataset
train_dataset, test_dataset = train_test_split(iris_dataset)
test_dataset.X
```

```python
# knn classifier
k = 2
kmeans = KNNClassifier(k)
kmeans.fit(train_dataset)
predictions = kmeans.predict(test_dataset)
predictions
```

```python
kmeans.score(test_dataset)
```

```python
# Stratified split dataset
strat_train_dataset, strat_test_dataset = stratified_train_test_split(iris_dataset, test_size=0.2, random_state=42)
print(strat_train_dataset.shape(), strat_test_dataset.shape())
```

```python
# Number of samples of each class in the train and test sets
import numpy as np

for name, dataset in (('train', strat_train_dataset), ('test', strat_test_dataset)):
    classes, counts = np.unique(dataset.y, return_counts=True)
    print(name, {str(c): int(n) for c, n in zip(classes, counts)})
```

```python
# knn classifier with the stratified split
knn_strat = KNNClassifier(k=3)
knn_strat.fit(strat_train_dataset)
knn_strat.score(strat_test_dataset)
```

```python
# cpu dataset (regression)
cpu_dataset = read_csv('../datasets/cpu/cpu.csv', features=True, label=True)
cpu_dataset.shape()
```

```python
# cpu features
cpu_dataset.features
```

```python
# Split dataset
cpu_train_dataset, cpu_test_dataset = train_test_split(cpu_dataset, test_size=0.2, random_state=42)
```

```python
# knn regressor
knn_regressor = KNNRegressor(k=3)
knn_regressor.fit(cpu_train_dataset)
cpu_predictions = knn_regressor.predict(cpu_test_dataset)
cpu_predictions
```

```python
# Real values (left) and predicted values (right)
np.c_[cpu_test_dataset.y, cpu_predictions][:5]
```

```python
# rmse between the predicted and the real values
knn_regressor.score(cpu_test_dataset)
```
